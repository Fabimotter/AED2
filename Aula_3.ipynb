Exercício 4 – Buscar cliente

```python

def buscar_cliente(nome, lista_clientes):
    for i in range(len(lista_clientes)):
        if lista_clientes[i] == nome:
            return i
    return -1


# Lista de clientes
clientes = ["Ana", "Carlos", "Maria", "Pedro", "Julia"]

# Testando um cliente que existe
resultado = buscar_cliente("Maria", clientes)

if resultado != -1:
    print(f"Cliente encontrado na posição {resultado}")
else:
    print("Cliente não encontrado")


# Testando um cliente que não existe
resultado = buscar_cliente("Fernanda", clientes)

if resultado != -1:
    print(f"Cliente encontrado na posição {resultado}")
else:
    print("Cliente não encontrado")
```

Exercício 5

```python
def conta_ocorrencias(valor, lista):
    contador = 0

    for elemento in lista:
        if elemento == valor:
            contador += 1

    return contador


# Casos de teste
print(conta_ocorrencias(3, [1, 3, 5, 3, 7]))   # 2
print(conta_ocorrencias(5, [5, 5, 5, 2, 1]))   # 3
print(conta_ocorrencias(8, [1, 2, 3, 4]))      # 0
print(conta_ocorrencias(2, []))                # 0
```

Exercicio 6

```python
def remove_elemento(lista, valor):
    for i in range(len(lista)):
        if lista[i] == valor:
            removido = lista[i]

            # Deslocar os elementos posteriores uma posição à esquerda
            for j in range(i, len(lista) - 1):
                lista[j] = lista[j + 1]

            lista.pop()  # Remove a última posição, agora repetida
            return removido

    return None


numeros = [10, 20, 30, 20, 40]
print("Antes:", numeros)
print("Removido:", remove_elemento(numeros, 20))
print("Depois:", numeros)  # [10, 30, 20, 40]
print("Valor ausente:", remove_elemento(numeros, 99))  # None
lista_vazia = []
print("Lista vazia:", remove_elemento(lista_vazia, 10))  # None
```

Projeto 1

```python
filmes = [
    "Titanic",
    "Avatar",
    "O Rei Leão",
    "Interestelar",
    "Barbie",
    "Matrix"
]


def buscar_filme(titulo, lista_filmes):
    for i in range(len(lista_filmes)):
        if lista_filmes[i].casefold() == titulo.strip().casefold():
            return i
    return -1


print("=== CATÁLOGO DE FILMES ===")
for filme in filmes:
    print("-", filme)

titulo = input("Digite o título do filme que deseja buscar: ")
posicao = buscar_filme(titulo, filmes)

if posicao != -1:
    print(f"Filme '{filmes[posicao]}' encontrado no índice {posicao}.")
else:
    print(f"O filme '{titulo.strip()}' não está no catálogo.")
```

```python
convidados = [
    "Ana",
    "Bruno",
    "Carla",
    "Daniel",
    "Eduarda",
    "Felipe"
]


def verificar_convidado(nome, lista_convidados):
    for convidado in lista_convidados:
        if convidado.casefold() == nome.strip().casefold():
            return True
    return False


print("=== VERIFICAÇÃO DE CONVIDADOS ===")
nome = input("Digite o nome do convidado: ")

if verificar_convidado(nome, convidados):
    print(f"{nome.strip()} está na lista de convidados. Entrada autorizada!")
else:
    print(f"{nome.strip()} não está na lista de convidados.")
```
