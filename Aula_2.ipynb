Projeto 1

```python
# Banco de dados fictício: cada ex-aluno é representado por um dicionário.
ex_alunos = [
    {"nome": "Ana Souza", "cpf": "00000000001", "curso": "Análise e Desenvolvimento de Sistemas"},
    {"nome": "Bruno Lima", "cpf": "00000000002", "curso": "Gestão Empresarial"},
    {"nome": "Carla Dias", "cpf": "00000000003", "curso": "Ciência de Dados"},
    {"nome": "Diego Alves", "cpf": "00000000004", "curso": "Logística"},
]

# Abordagem 1 — busca linear: O(n) no pior caso.
def buscar_aluno_linear(lista_alunos, cpf_buscado):
    for aluno in lista_alunos:
        if aluno["cpf"] == cpf_buscado:
            return aluno
    return None

# Pré-processamento: O(n) de tempo e O(n) de memória.
def criar_indice_cpfs(lista_alunos):
    return {aluno["cpf"]: aluno for aluno in lista_alunos}

# Abordagem 2 — busca por dicionário: O(1) em média após pré-processar.
def buscar_aluno_eficiente(indice_cpfs, cpf_buscado):
    return indice_cpfs.get(cpf_buscado)

cpf_procurado = "00000000003"
indice_ex_alunos = criar_indice_cpfs(ex_alunos)

print("Resultado da busca linear:")
print(buscar_aluno_linear(ex_alunos, cpf_procurado))
print("\nResultado da busca eficiente:")
print(buscar_aluno_eficiente(indice_ex_alunos, cpf_procurado))
print("\nCPF não cadastrado:", buscar_aluno_eficiente(indice_ex_alunos, "99999999999"))
```

Conclusão: para localizar muitos ex-alunos, indexar uma vez e consultar o dicionário é mais eficiente. Uma única consulta com construção do índice também tem custo total O(n); a vantagem aparece quando o índice é reutilizado. Se o mesmo CPF aparecer mais de uma vez, o índice manterá o último registro (o enunciado pressupõe CPF único

```python
def recomendar_disciplinas(historicos, historico_aluno):
    disciplinas_aluno = set(historico_aluno)
    recomendacoes = []
    ja_recomendadas = set()

    for historico_outro in historicos:
        disciplinas_outro = set(historico_outro)

        # Só recomenda se existir ao menos uma disciplina em comum.
        if disciplinas_aluno.intersection(disciplinas_outro):
            for disciplina in historico_outro:
                if disciplina not in disciplinas_aluno and disciplina not in ja_recomendadas:
                    recomendacoes.append(disciplina)
                    ja_recomendadas.add(disciplina)

    return recomendacoes

# Históricos de outros alunos (códigos fictícios de disciplinas).
historicos_outros = [
    ["ALGO", "POO", "IA"],
    ["REDES", "SEGURANCA"],
    ["BD", "WEB", "IA"],
    ["CALCULO", "ESTATISTICA"],
    ["ALGO", "ESTRUTURA_DADOS"],
]

historico_aluno = ["ALGO", "BD"]
recomendacoes = recomendar_disciplinas(historicos_outros, historico_aluno)

print("Disciplinas já cursadas:", historico_aluno)
print("Disciplinas recomendadas:", recomendacoes)
```
