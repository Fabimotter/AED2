```python
class Livro:
  def __init__(self, titulo, autor, paginas):
    self.titulo = titulo
    self.autor = autor
    self.paginas = paginas

  def descricao(self):
    return f"{self.titulo} - {self.autor} ({self.paginas} páginas)"

if __name__ == "__main__":
  livro1 = Livro("Dom Casmurro", "Machado de Assis", 200)
  livro2 = Livro("O Alquimista", "Paulo Coelho", 200)

  print(livro1.descricao())
  print(livro2.descricao())
```

```python
class Conta:
  def __init__(self, titular):
    self.titular = titular
    self.__saldo = 0

  def depositar(self, valor):
    if valor <= 0:
      print("Valor inválido")
      return False
    else:
      self.__saldo += valor
      return True

  def consultar_saldo(self):
    return self.__saldo

if __name__ == "__main__":
  conta = Conta("Ana Souza")

  conta.depositar(150.00)
  conta.depositar(70.50)

  print(f"Titular: {conta.titular}")
  print(f"Saldo atual: R$ {conta.consultar_saldo():.2f}")
```

```python
class Animal:
  def __init__(self, nome):
    self.nome = nome

  def emitir_som(self):
    return "Som genérico"

class Cachorro(Animal):
  def emitir_som(self):
    return "Au Au"

class Gato(Animal):
  def emitir_som(self):
    return "Miau"

if __name__ == "__main__":
    animais = [Cachorro("Rex"), Gato("Mimi")]

    # Mesma chamada para todos, sem testar o tipo: cada objeto responde do seu jeito
    for animal in animais:
        print(f"{animal.nome}: {animal.emitir_som()}")
```

```python
# Exercício 7: Veículo com velocidade controlada


class Veiculo:
    def __init__(self, placa, velocidade_max):
        self.__placa = placa
        self.__velocidade_max = velocidade_max
        self.__velocidade_atual = 0

    # Getters
    def get_placa(self):
        return self.__placa

    def get_velocidade_max(self):
        return self.__velocidade_max

    def get_velocidade_atual(self):
        return self.__velocidade_atual

    # Setter validado
    def set_velocidade_max(self, nova_max):
        if nova_max <= 0:
            print("Alteração recusada: a velocidade máxima deve ser positiva.")
            return False
        if nova_max < self.__velocidade_atual:
            print("Alteração recusada: a nova máxima é menor que a velocidade atual.")
            return False
        self.__velocidade_max = nova_max
        return True

    def acelerar(self, incremento=10):
       # min() garante que a velocidade nunca ultrapassa a máxima
        self.__velocidade_atual = min(
            self.__velocidade_atual + incremento, self.__velocidade_max
        )

    def frear(self, decremento=10):
        # max() garante que a velocidade nunca fica negativa
        self.__velocidade_atual = max(self.__velocidade_atual - decremento, 0)

    def __str__(self):
        return (
            f"Veículo {self.__placa} | velocidade atual: "
            f"{self.__velocidade_atual} km/h | máxima: {self.__velocidade_max} km/h"
        )


if __name__ == "__main__":
    carro = Veiculo("ABC1D23", 120)
    print(carro)

    carro.acelerar()  # usa o incremento padrão (10)
    carro.acelerar(50)
    print(carro)

    carro.acelerar(200)  # tenta passar da máxima: fica em 120
    print(carro)

    carro.set_velocidade_max(100)  # recusado: menor que a velocidade atual
    carro.set_velocidade_max(-10)  # recusado: valor não positivo

    carro.frear()  # usa o decremento padrão (10)
    carro.frear(30)
    print(carro)

    carro.set_velocidade_max(100)  # agora é aceito (atual = 80)
    print(carro)

    carro.frear(500)  # tenta ficar negativa: para em 0
    print(carro)

    print(f"Placa: {carro.get_placa()} | Máxima: {carro.get_velocidade_max()} km/h")
```

```python
# Exercício 8: Funcionários e cálculo de bônus


class Funcionario:
    def __init__(self, nome, salario):
        self.nome = nome
        self.salario = salario

    def calcular_bonus(self):
        return self.salario * 0.05

    def __str__(self):
        return f"{self.nome} (salário: R$ {self.salario:.2f})"


class Vendedor(Funcionario):
    def __init__(self, nome, salario, vendas):
        super().__init__(nome, salario)
        self.vendas = vendas

    def calcular_bonus(self):
        # 5% do salário + 2% das vendas
        return self.salario * 0.05 + self.vendas * 0.02


class Gerente(Funcionario):
    def __init__(self, nome, salario, equipe):
        super().__init__(nome, salario)
        self.equipe = equipe  # quantidade de integrantes da equipe

    def calcular_bonus(self):
        # 10% do salário + R$ 100 por integrante da equipe
        return self.salario * 0.10 + self.equipe * 100


if __name__ == "__main__":
    funcionarios = [
        Funcionario("Carlos", 3000.00),
        Vendedor("Marina", 2500.00, 40000.00),
        Gerente("Paulo", 8000.00, 6),
    ]

    # Mesmo método para todos; cada classe calcula o bônus à sua maneira
    for funcionario in funcionarios:
        tipo = type(funcionario).__name__
        bonus = funcionario.calcular_bonus()
        print(f"{funcionario.nome} | {tipo} | bônus: R$ {bonus:.2f}")
```

```python

class Produto:
    contador = 0  # atributo de classe: compartilhado por todos os objetos

    def __init__(self, nome, preco):
        Produto.contador += 1
        self.__id = Produto.contador  # id sequencial automático
        self.__nome = nome
        self.__preco = preco

    # Getters
    def get_id(self):
        return self.__id

    def get_nome(self):
        return self.__nome

    def get_preco(self):
        return self.__preco

    # Setter validado
    def set_preco(self, novo_preco):
        if novo_preco <= 0:
            print("Preço recusado: o valor deve ser maior que zero.")
            return False
        self.__preco = novo_preco
        return True

    def aplicar_desconto(self, percentual):
        if percentual < 0 or percentual > 100:
            print("Desconto recusado: o percentual deve estar entre 0 e 100.")
            return False
        self.__preco = self.__preco * (1 - percentual / 100)
        return True

    def __str__(self):
        return f"[{self.__id}] {self.__nome} | R$ {self.__preco:.2f}"


if __name__ == "__main__":
    p1 = Produto("Teclado mecânico", 350.00)
    p2 = Produto("Mouse sem fio", 120.00)
    p3 = Produto("Monitor 24 polegadas", 900.00)

    p1.aplicar_desconto(10)  # 350,00 passa a 315,00
    p2.set_preco(135.00)
    p3.set_preco(-50)  # recusado
    p3.aplicar_desconto(150)  # recusado

    for produto in (p1, p2, p3):
        print(produto)

    print(f"Produtos criados: {Produto.contador}")
```

```python
# Exercício 10: Mini sistema de pedidos de e-commerce


class Produto:
    contador = 0  # atributo de classe usado para gerar o id automático

    def __init__(self, nome, preco):
        if preco <= 0:
            raise ValueError("O preço deve ser maior que zero.")
        Produto.contador += 1
        self.__id = Produto.contador
        self.__nome = nome
        self.__preco = preco

    def get_id(self):
        return self.__id

    def get_nome(self):
        return self.__nome

    def get_preco(self):
        return self.__preco

    def set_preco(self, novo_preco):
        if novo_preco <= 0:
            print("Preço recusado: o valor deve ser maior que zero.")
            return False
        self.__preco = novo_preco
        return True

    def calcular_frete(self):
        return 0.0

    def __str__(self):
        return f"[{self.__id}] {self.__nome} | R$ {self.__preco:.2f}"


class ProdutoFisico(Produto):
    def __init__(self, nome, preco, peso):
        super().__init__(nome, preco)
        self.__peso = peso  # em kg

    def get_peso(self):
        return self.__peso

    def calcular_frete(self):
        # R$ 5,00 fixos + R$ 2,00 por kg
        return 5.00 + 2.00 * self.__peso

    def __str__(self):
        return f"{super().__str__()} | físico, {self.__peso} kg"


class ProdutoDigital(Produto):
    def __init__(self, nome, preco, tamanho_mb):
        super().__init__(nome, preco)
        self.__tamanho_mb = tamanho_mb

    def get_tamanho_mb(self):
        return self.__tamanho_mb

    # calcular_frete() não é redefinido: herda o frete 0 de Produto

    def __str__(self):
        return f"{super().__str__()} | digital, {self.__tamanho_mb} MB"


class Cliente:
    def __init__(self, nome, email):
        self.__nome = nome
        self.__email = email

    def get_nome(self):
        return self.__nome

    def get_email(self):
        return self.__email

    def __str__(self):
        return f"{self.__nome} <{self.__email}>"


class Pedido:
    contador = 0  # atributo de classe usado para gerar o número automático

    def __init__(self, cliente):
        Pedido.contador += 1
        self.__numero = Pedido.contador
        self.__cliente = cliente
        self.__produtos = []
        self.__status = "aberto"

    def get_numero(self):
        return self.__numero

    def get_status(self):
        return self.__status

    def adicionar_produto(self, produto):
        # A regra do pedido fechado é controlada aqui dentro, não por quem usa a classe
        if self.__status == "fechado":
            print(
                f"Pedido {self.__numero} está fechado: "
                f"não foi possível adicionar '{produto.get_nome()}'."
            )
            return False
        self.__produtos.append(produto)
        return True

    def calcular_total(self):
        return sum(produto.get_preco() for produto in self.__produtos)

    def calcular_frete_total(self):
        # Polimorfismo: a mesma chamada serve para produto físico e digital
        return sum(produto.calcular_frete() for produto in self.__produtos)

    def fechar_pedido(self):
        if self.__status == "fechado":
            print(f"Pedido {self.__numero} já estava fechado.")
            return False
        if not self.__produtos:
            print(f"Pedido {self.__numero} não pode ser fechado sem produtos.")
            return False
        self.__status = "fechado"
        return True

    def __str__(self):
        linhas = [
            f"Pedido nº {self.__numero} ({self.__status})",
            f"Cliente: {self.__cliente}",
            "Itens:",
        ]
        for produto in self.__produtos:
            linhas.append(f"  {produto} | frete: R$ {produto.calcular_frete():.2f}")
        total = self.calcular_total()
        frete = self.calcular_frete_total()
        linhas.append(f"Total dos produtos: R$ {total:.2f}")
        linhas.append(f"Frete total: R$ {frete:.2f}")
        linhas.append(f"Total a pagar: R$ {total + frete:.2f}")
        return "\n".join(linhas)


if __name__ == "__main__":
    cliente = Cliente("Ana Souza", "ana.souza@email.com")

    notebook = ProdutoFisico("Notebook", 3500.00, 2.5)
    ebook = ProdutoDigital("E-book de Python", 59.90, 12)
    mochila = ProdutoFisico("Mochila", 180.00, 0.8)
    curso = ProdutoDigital("Curso de POO", 199.00, 1500)

    pedido = Pedido(cliente)
    pedido.adicionar_produto(notebook)
    pedido.adicionar_produto(ebook)
    pedido.adicionar_produto(mochila)

    print(pedido)
    print()

    pedido.fechar_pedido()
    print(f"Status depois de fechar: {pedido.get_status()}")

    # Tentativa de adicionar depois do fechamento: a própria classe recusa
    pedido.adicionar_produto(curso)
    print()
    print(pedido)
```
