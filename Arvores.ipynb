Exercicio 1

```python
# Cada nó pode ter um filho à esquerda e outro à direita
class No:
    def __init__(self, valor, esquerda=None, direita=None):
        self.valor = valor
        self.esquerda = esquerda
        self.direita = direita


def altura(no):
    # Uma árvore vazia tem altura 0
    if no is None:
        return 0

    altura_esquerda = altura(no.esquerda)
    altura_direita = altura(no.direita)

    return 1 + max(altura_esquerda, altura_direita)


# Árvore de exemplo com três níveis
raiz = No("A")
raiz.esquerda = No("B")
raiz.direita = No("C")
raiz.esquerda.esquerda = No("D")

print("Altura da árvore:", altura(raiz))
```

```python
class No:
    def __init__(self, valor, esquerda=None, direita=None):
        self.valor = valor
        self.esquerda = esquerda
        self.direita = direita


def conta_folhas(raiz):
    if raiz is None:
        return 0

    if raiz.esquerda is None and raiz.direita is None:
        return 1

    return conta_folhas(raiz.esquerda) + conta_folhas(raiz.direita)


# Árvore de exemplo
raiz = No(
    1,
    esquerda=No(2, No(4), No(5)),
    direita=No(3, direita=No(6))
)

print(conta_folhas(raiz))  # Saída: 3
```

```python
from sklearn import tree

# Cada lista em X contém [idade, renda].
# Em y, 0 = ruim pagador e 1 = bom pagador.
X = [[25, 5000], [40, 12000], [22, 3000], [35, 8000], [50, 10000]]
y = [1, 1, 0, 1, 0]

# Cria e treina a árvore de decisão
clf = tree.DecisionTreeClassifier()
clf = clf.fit(X, y)

# Prevê a classificação de um cliente de 30 anos com renda de 7000
print(clf.predict([[30, 7000]]))
```

```python
# Define a estrutura familiar (árvore genealógica simples)
familia = {
    'João': ['Pedro', 'Ana'],
    'Pedro': ['Carlos', 'Marina'],
    'Ana': ['Julia'],
    'Carlos': [],
    'Marina': [],
    'Julia': []
}

# Exibe a árvore a partir de um nome, recursivamente
def exibir_arvore(nome, nivel=0):
    print("  " * nivel + nome)
    for filho in familia.get(nome, []):
        exibir_arvore(filho, nivel + 1)

# Exemplo de uso
exibir_arvore('João')
```
