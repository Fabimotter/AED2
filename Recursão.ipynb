Questão 5

```python
def soma_recursiva(n):
    if n == 0:          # Caso base: encerra a recursão
        return 0
    return n + soma_recursiva(n - 1)

print(soma_recursiva(5))
```

```python
def contar_funcionarios(departamento):
    # Começa pelos funcionários diretos
    total = departamento.get("funcionarios", 0)

    # Soma os funcionários de cada subdepartamento
    for subdep in departamento.get("subdepartamentos", []):
        total += contar_funcionarios(subdep)

    return total


estrutura = {
    "funcionarios": 10,
    "subdepartamentos": [
        {
            "funcionarios": 5,
            "subdepartamentos": []
        },
        {
            "funcionarios": 8,
            "subdepartamentos": [
                {"funcionarios": 3, "subdepartamentos": []},
                {"funcionarios": 2, "subdepartamentos": []}
            ]
        }
    ]
}

print("Total de funcionários:", contar_funcionarios(estrutura))
```

```python
import os
import tempfile

def encontrar_arquivos(diretorio, extensao):
    arquivos_encontrados = []

    for item in os.listdir(diretorio):
        caminho_completo = os.path.join(diretorio, item)

        if os.path.isfile(caminho_completo):
            if item.lower().endswith(extensao.lower()):
                arquivos_encontrados.append(os.path.abspath(caminho_completo))

        elif os.path.isdir(caminho_completo):
            # Pesquisa também dentro da subpasta
            arquivos_encontrados.extend(
                encontrar_arquivos(caminho_completo, extensao)
            )

    return arquivos_encontrados


# Exemplo para testar no Colab
with tempfile.TemporaryDirectory() as pasta_exemplo:
    subpasta = os.path.join(pasta_exemplo, "relatorios")
    os.mkdir(subpasta)

    for caminho in [
        os.path.join(pasta_exemplo, "leia-me.txt"),
        os.path.join(subpasta, "resumo.txt"),
        os.path.join(subpasta, "dados.csv")
    ]:
        with open(caminho, "w", encoding="utf-8") as arquivo:
            arquivo.write("Exemplo")

    print("Arquivos .txt encontrados:")
    for arquivo in encontrar_arquivos(pasta_exemplo, ".txt"):
        print(arquivo)
```
