```python
#grafo de exemplo
G={'A':['B','C'],
   'B':['A','D','E'],
   'C':['A','F',],
   'D':['B'],
   'E': ['B','F'],
   'F':['C','E']}
```

```python
#funÇÂO DFS recursiva
def dfs(grafo, inicio, visitados=None):
    if visitados is None:
        visitados = []
    visitados.append(inicio)
    for vizinho in grafo[inicio]:
        if vizinho not in visitados:
            dfs(grafo, vizinho, visitados)
    return visitados

ordem = dfs(G, 'A')
print(ordem)
```

```python
#grafo de exemplo
G={'A':['B','C'],
   'B':['A','D','E'],
   'C':['A','F',],
   'D':['B'],
   'E': ['B','F'],
   'F':['C','E']}
```

```python
from collections import deque
def bfs(grafo, inicio):
    visitados = [inicio]
    fila=deque([inicio])
    while fila:
        v = fila.popleft()
        ordem.append(v)
        for vizinho in grafo[v]:
            if vizinho not in visitados:
                visitados.append(vizinho)
                fila.append(vizinho)
    return ordem
    ordem_bfs = bfs(G, 'A')
    print(ordem_bfs)
```

```python
import networkx as nx

# Criação do grafo ponderado
G = nx.Graph()
G.add_weighted_edges_from([
    ('São Paulo', 'Campinas', 96),
    ('Campinas', 'São Paulo', 95),
    ('Campinas', 'Sorocaba', 85),
    ('Sorocaba', 'Campinas', 85),
    ('Campinas', 'Ribeirão Preto', 238),
    ('Ribeirão Preto', 'Campinas', 238),
    ('Ribeirão Preto', 'Barretos', 127),
    ('Barretos', 'Ribeirão Preto', 127)
    # (pode-se adicionar outras arestas de exemplo conforme os dados)
])

# Cálculo do menor caminho e sua distância
origem = 'São Paulo'
destino = 'Barretos'
menor_caminho = nx.shortest_path(G, origem, destino, weight='weight')
distancia = nx.shortest_path_length(G, origem, destino, weight='weight')
print("Menor caminho de {} até {}:".format(origem, destino), menor_caminho)
print("Distância total:", distancia, "km")
```

```python
from collections import deque

# Grafo representando amizades
rede = {
    'A': ['B', 'D'],
    'B': ['A', 'C'],
    'C': ['B', 'D', 'E'],
    'D': ['A', 'C', 'F'],
    'E': ['C'],
    'F': ['D']
}

def bfs_caminho(grafo, inicio, destino):
    visitados = {inicio}
    fila = deque([(inicio, [inicio])])

    while fila:
        vertice, caminho = fila.popleft()
        if vertice == destino:
            return caminho

        for viz in grafo[vertice]:
            if viz not in visitados:
                visitados.add(viz)
                fila.append((viz, caminho + [viz]))

    return None

# Exemplo: encontrar caminho de A até E
orig = 'A'
dest = 'E'
caminho = bfs_caminho(rede, orig, dest)
print("Caminho mínimo de '{}' até '{}':".format(orig, dest),
      caminho)
```
