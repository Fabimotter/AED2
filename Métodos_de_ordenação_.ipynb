```python
# Faturamento de janeiro a dezembro, em milhares de reais
vendas = [120, 85, 90, 100, 75, 130, 95, 110, 80, 140, 70, 105]

def particao_decrescente(arr, low, high):
    pivot = arr[high]
    i = low - 1

    for j in range(low, high):
        # Coloca os valores maiores à esquerda do pivô
        if arr[j] >= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

def quick_sort_decrescente(arr, low=0, high=None):
    if high is None:
        high = len(arr) - 1

    if low < high:
        pi = particao_decrescente(arr, low, high)
        quick_sort_decrescente(arr, low, pi - 1)
        quick_sort_decrescente(arr, pi + 1, high)

    return arr

# A cópia preserva a lista original
vendas_ordenadas = quick_sort_decrescente(vendas.copy())

print("Vendas mensais (original):", vendas)
print("Vendas mensais ordenadas (decrescente):", vendas_ordenadas)
print("Maior faturamento:", vendas_ordenadas[0], "mil reais")
print("Menor faturamento:", vendas_ordenadas[-1], "mil reais")
```

```python
import random

# Gera 5 cartas diferentes, com valores de 1 a 99
cartas = random.sample(range(1, 100), 5)

# A última posição da lista será o topo da pilha
pilha_compras = cartas.copy()
random.shuffle(pilha_compras)

mao = []

print("Cartas na pilha (do topo para a base):", pilha_compras[::-1])

while pilha_compras:
    input("\nPressione Enter para comprar uma carta...")

    carta = pilha_compras.pop()
    print(f"Você comprou a carta: {carta}")

    # Procura onde a carta deve entrar na mão
    pos = 0
    while pos < len(mao) and mao[pos] < carta:
        pos += 1

    mao.insert(pos, carta)
    print("Sua mão agora:", mao)


print("\nTodas as cartas foram compradas e ordenadas na mão!")
```
