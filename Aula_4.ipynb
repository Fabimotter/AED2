Exercicio 3

```python
def busca_binaria(vetor, alvo, esq=0, dir=None):
    if dir is None:
        dir = len(vetor) - 1

    # Caso-base: intervalo vazio, elemento não encontrado.
    if esq > dir:
        return -1

    meio = (esq + dir) // 2

    if vetor[meio] == alvo:
        return meio
    elif alvo < vetor[meio]:
        return busca_binaria(vetor, alvo, esq, meio - 1)
    else:
        return busca_binaria(vetor, alvo, meio + 1, dir)


numeros = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]

print("Lista ordenada:", numeros)
print("Índice de 23:", busca_binaria(numeros, 23))
print("Índice de 50:", busca_binaria(numeros, 50))
```

Exercicio 4

```python
def busca_sequencial_com_contagem(vetor, alvo):
    comparacoes = 0

    for indice, valor in enumerate(vetor):
        comparacoes += 1
        if valor == alvo:
            return indice, comparacoes

    return -1, comparacoes


def busca_binaria_com_contagem(vetor, alvo):
    esq = 0
    dir = len(vetor) - 1
    comparacoes = 0

    while esq <= dir:
        meio = (esq + dir) // 2
        comparacoes += 1

        if vetor[meio] == alvo:
            return meio, comparacoes
        elif alvo < vetor[meio]:
            dir = meio - 1
        else:
            esq = meio + 1

    return -1, comparacoes


lista = list(range(0, 2000, 2))  # 1.000 inteiros pares em ordem crescente
alvo = 2001                      # Número ausente da lista

indice_seq, comp_seq = busca_sequencial_com_contagem(lista, alvo)
indice_bin, comp_bin = busca_binaria_com_contagem(lista, alvo)

print("Tamanho da lista:", len(lista))
print("Número procurado:", alvo)
print("Busca sequencial - índice:", indice_seq, "| comparações:", comp_seq)
print("Busca binária    - índice:", indice_bin, "| comparações:", comp_bin)
```

Projeto 1

```python
produtos = [
    {"nome": "Notebook", "preco": 3500.00, "descricao": "Notebook com 16 GB de RAM"},
    {"nome": "Mouse", "preco": 85.90, "descricao": "Mouse sem fio"},
    {"nome": "Teclado", "preco": 150.00, "descricao": "Teclado mecânico"},
    {"nome": "Monitor", "preco": 1200.00, "descricao": "Monitor de 24 polegadas"},
    {"nome": "Webcam", "preco": 210.00, "descricao": "Webcam Full HD"},
]

# A busca binária exige a lista ordenada pela mesma chave usada na busca.
produtos = sorted(produtos, key=lambda produto: produto["nome"].casefold())


def buscar_produto(catalogo, nome):
    esq = 0
    dir = len(catalogo) - 1
    alvo = nome.strip().casefold()

    while esq <= dir:
        meio = (esq + dir) // 2
        nome_atual = catalogo[meio]["nome"].casefold()

        if nome_atual == alvo:
            return catalogo[meio]
        elif alvo < nome_atual:
            dir = meio - 1
        else:
            esq = meio + 1

    return None


nome_procurado = "Notebook"  # Experimente "Mouse" ou "Impressora".
produto = buscar_produto(produtos, nome_procurado)

if produto is not None:
    print("Produto encontrado!")
    print("Nome:", produto["nome"])
    print(f"Preço: R$ {produto['preco']:.2f}")
    print("Descrição:", produto["descricao"])
else:
    print("Produto não encontrado no catálogo.")
```

```python
livros = [
    {"titulo": "Dom Casmurro", "autor": "Machado de Assis", "ano": 1899},
    {"titulo": "O Pequeno Príncipe", "autor": "Antoine de Saint-Exupéry", "ano": 1943},
    {"titulo": "1984", "autor": "George Orwell", "ano": 1949},
    {"titulo": "Capitães da Areia", "autor": "Jorge Amado", "ano": 1937},
    {"titulo": "A Hora da Estrela", "autor": "Clarice Lispector", "ano": 1977},
]

livros = sorted(livros, key=lambda livro: livro["titulo"].casefold())


def buscar_livro(indice_livros, titulo):
    esq = 0
    dir = len(indice_livros) - 1
    alvo = titulo.strip().casefold()

    while esq <= dir:
        meio = (esq + dir) // 2
        titulo_atual = indice_livros[meio]["titulo"].casefold()

        if titulo_atual == alvo:
            return indice_livros[meio]
        elif alvo < titulo_atual:
            dir = meio - 1
        else:
            esq = meio + 1

    return None


titulo_procurado = "Dom Casmurro"  # Experimente "1984" ou "Livro inexistente".
livro = buscar_livro(livros, titulo_procurado)

if livro is not None:
    print("Livro encontrado!")
    print("Título:", livro["titulo"])
    print("Autor:", livro["autor"])
    print("Ano de publicação:", livro["ano"])
else:
    print("Livro não encontrado na biblioteca.")
```
