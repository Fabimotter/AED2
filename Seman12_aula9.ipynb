Exercício 3

```python
def soma_digitos (n):
    if n < 10: # Caso base: número com um único dígito
        return n
    else:
        ultimo_digito = n % 10
        resto_do_numero = n // 10
        return ultimo_digito + soma_digitos ( resto_do_numero ) # Passo recursivo

print( soma_digitos (123))
```

Exercício 4

```python
def palindromo(s):
    # Normaliza: remove espaços e ignora maiúsculas/minúsculas
    s = s.lower().replace(" ", "")

    # Caso base: 0 ou 1 caractere é sempre palíndromo
    if len(s) <= 1:
        return True

    # Compara o primeiro e o último caractere
    elif s[0] == s[-1]:
        # Passo recursivo: verifica o miolo da string
        return palindromo(s[1:-1])

    else:
        return False


print(palindromo("A man a plan a canal Panama"))
print(palindromo("hello"))
```

Exercicio 5 Função iterativa

```python
def soma_iterativa(n):
  soma=0
  for i in range(n+1):
    soma+=i
  return soma

print(soma_iterativa(5))
```

Exercicio 5 Função recursiva

```python
def soma_recursiva(n):
  if n==1: #casobase
    return 1
  else:
    return n + soma_recursiva(n-1)

print(soma_recursiva(5))
```
